## Question 1 — Hypothesis Testing (25 marks)

A pharmaceutical company claims that its antibiotic is effective in **at least 80%** of cases. A hospital audit of $n = 20$ randomly selected patients who received the antibiotic found that it was effective in only $x = 11$ cases.

Test whether the audit provides sufficient evidence to doubt the company’s claim at the **5% significance level**.

**(a)** State $H_0$ and $H_1$, and identify the test as left-tailed, right-tailed, or two-tailed. [4 marks]

**(b)** Let $X$ be the number of patients for whom the antibiotic is effective. State the distribution of $X$ under $H_0$. Explain why the claimed boundary $p = 0.80$ is used for the calculation. [5 marks]

**(c)** Calculate the one-tailed $p$-value $P(X \leq 11)$ under $X \sim \operatorname{Bin}(20, 0.80)$. You may build the lower-tail probability from these approximate components:

- $P(X = 11) \approx 0.0074$
- $P(X = 10) \approx 0.0020$
- $P(X = 9) \approx 0.0005$
- $P(X \leq 8) \approx 0.0001$

[7 marks]

**(d)** Compare the $p$-value with $\alpha = 0.05$. State a correct statistical decision and a one-sentence conclusion for a hospital administrator. [6 marks]

**(e)** Write Python code to verify $P(X \leq 11)$ by summing the Binomial probabilities from $k=0$ to $11$. [3 marks]

```
WORKING:

````python
# Q1e -- exact binomial test: antibiotic effectiveness
from math import comb

n = 20
p_null = 0.80
x_observed = 11

# Lower-tail p-value for H1: p < 0.80.
def binomial_pmf(k, trials, probability):
    return comb(trials, k) * probability**k * (1 - probability)**(trials - k)

p_value = sum(binomial_pmf(k, n, p_null) for k in range(x_observed + 1))

print(f"Under H0: X ~ Bin({n}, {p_null:.2f})")
print(f"Observed effective cases: {x_observed} out of {n}")
print(f"p-value = P(X <= {x_observed}) = {p_value:.4f}")
print(f"\nConclusion at alpha = 0.05: {'Reject H0' if p_value < 0.05 else 'Fail to reject H0'}")
print("The audit provides evidence that effectiveness is below the claimed 80%.")
````

**(c)** Verify $x^*$ is a maximum using the second derivative test. [4 marks]

````python
# Q2 — Dairy profit optimisation
def C(x):
    return 2*x**2 - 80*x + 1200

def Revenue(x):
    return 1200 * x   # $1200 per thousand litres

def Pi(x):
    return Revenue(x) - C(x)

def dPi(x):
    return 1200 - (4*x - 80)   # Pi'(x) = R'(x) - C'(x)

# Critical point: dPi(x) = 0 → 1200 - 4x + 80 = 0 → 4x = 1280 → x* = 320
x_star = (1200 + 80) / 4
print(f"x* = {x_star}")
print(f"Pi''(x) = -4 < 0 → maximum confirmed")
print(f"Pi(x*) = ${Pi(x_star):,.2f}")

# (e) Fuel costs increase: fixed cost rises from 1200 to 1500
def C_new(x):
    return 2*x**2 - 80*x + 1500

def Pi_new(x):
    return Revenue(x) - C_new(x)

print(f"\nWith higher fixed cost (C0=1500):")
print(f"Pi_new'(x) = 1200 - (4x - 80)  → same critical point, x* = {x_star} (unchanged,")
print(f"  since a constant cost shift doesn't affect the derivative)")
print(f"Pi_new(x*) = ${Pi_new(x_star):,.2f}  (${Pi(x_star) - Pi_new(x_star):,.0f} less than before)")
````

## Question 3 — Integration (25 marks)

A solar panel array generates power:

$$P(t) = 500 \sin\!\left(\frac{\pi t}{12}\right) \text{ watts}, \quad 0 \leq t \leq 12 \text{ hours}$$

**(a)** Sketch $P(t)$ over 12 hours. What is peak power and when does it occur? [4 marks]

**(b)** Set up the integral for total energy generated (watt-hours) over the full day. [4 marks]

**(c)** Evaluate $\displaystyle\int_0^{12} 500 \sin\!\left(\frac{\pi t}{12}\right)dt$ analytically. [10 marks]

**(d)** Convert to kWh. If electricity sells at 30 cents/kWh, what is daily revenue? [4 marks]

**(e)** Write Python code to verify the energy calculation numerically and calculate the corresponding daily revenue. [3 marks]

````python
# Q3 — Solar energy integration
def P(t):
    return 500 * np.sin(np.pi * t / 12)

# Analytic: ∫₀¹² 500 sin(πt/12) dt = 500 × [-12/π cos(πt/12)]₀¹²
# = 500 × (-12/π)(cos(π) - cos(0)) = 500 × (-12/π)(-1 - 1) = 500 × 24/π
energy_Wh_analytic = 500 * 24 / np.pi
print(f"Analytic energy: {energy_Wh_analytic:.2f} Wh = {energy_Wh_analytic/1000:.4f} kWh")

# Numerical verification
from numpy import trapz
t_fine = np.linspace(0, 12, 10000)
energy_Wh_num = trapz(P(t_fine), t_fine)
print(f"Numerical energy: {energy_Wh_num:.2f} Wh")

# Revenue
energy_kWh = energy_Wh_analytic / 1000
revenue = energy_kWh * 0.30
print(f"Daily revenue: ${revenue:.4f}")

t_vals = np.linspace(0, 12, 200)
fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(t_vals, P(t_vals), "goldenrod", lw=2.5)
ax.fill_between(t_vals, P(t_vals), alpha=0.2, color="goldenrod")
ax.set_xlabel("Hour of day"); ax.set_ylabel("Power (W)")
ax.set_title("Solar Panel Output P(t) = 500 sin(πt/12)")
ax.grid(alpha=0.3)
plt.tight_layout(); plt.show()
````

A water tank drains according to Torricelli's law:

$$\frac{dV}{dt} = -0.5\sqrt{V}, \quad V(0) = 100 \text{ L}$$

**(a)** Separate variables and integrate to find an explicit formula for $V(t)$. [10 marks]

**(b)** How long until the tank is half empty ($V=50$ L)? [5 marks]

**(c)** How long until the tank is completely empty ($V=0$ L)? [4 marks]

**(d)** Write Python code to plot your analytic formula for $0 \leq t \leq 40$, labelling the starting volume, half-empty point, and empty-tank time. [6 marks]

````python
# Q4 optional extension -- plot the analytic draining-tank solution
def volume(t):
    remaining_height = 10 - t / 4
    return np.maximum(remaining_height, 0)**2

half_empty_time = 40 * (1 - 1 / np.sqrt(2))
empty_time = 40

t_values = np.linspace(0, empty_time, 400)
fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(t_values, volume(t_values), color="steelblue", lw=2.5, label=r"$V(t)=(10-t/4)^2$")
ax.scatter([0, half_empty_time, empty_time], [100, 50, 0], color="crimson", zorder=3)
ax.annotate("Start: 100 L", (0, 100), xytext=(3, 88))
ax.annotate(f"Half empty: {half_empty_time:.1f} min", (half_empty_time, 50), xytext=(16, 62))
ax.annotate("Empty: 40 min", (empty_time, 0), xytext=(26, 14))
ax.set_xlabel("Time (min)")
ax.set_ylabel("Volume (L)")
ax.set_title("Tank Draining: Analytic Solution")
ax.legend()
ax.grid(alpha=0.3)
plt.tight_layout()
plt.show()
````
